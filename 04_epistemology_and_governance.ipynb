```python
# Optional T4: tiny NS-like advection on GPU (numerical demo, not Clay)
try:
    import torch
    if not torch.cuda.is_available():
        print(json.dumps({'gpu_ns': 'skipped'}))
    else:
        n = 128
        u = torch.randn(n, n, device='cuda') * 0.01
        nu = 0.08
        for _ in range(30):
            lap = (torch.roll(u,1,0)+torch.roll(u,-1,0)+torch.roll(u,1,1)+torch.roll(u,-1,1)-4*u)
            u = u + 0.05 * (nu * lap)
        E = float((0.5 * u.pow(2).mean()).cpu())
        print(json.dumps({'gpu': torch.cuda.get_device_name(0), 'E': E, 'watch': 'regular_so_far' if E < 1e8 else 'numerical_ns_blowup', 'not_clay': True}))
except Exception as e:
    print(json.dumps({'gpu_ns': str(e)}))
```

## 4. Gemini in Colab and VS Code

Same review in both places. The key is never written into this file.

**Colab:** Key icon → Add secret → name `GEMINI_API_KEY` → notebook access on. Runtime → Run all.

**VS Code:** `pip install -U google-genai`, then in the terminal `set GEMINI_API_KEY` (Windows) or `export GEMINI_API_KEY` (macOS/Linux). Run the next cell, or `python notebooks/gemini_review.py`.

Model: `gemini-3.8-flash` via `google-genai`. The review must keep measured, telemetry, and author_claim apart.


```python
# Gemini packet review. Colab secret or VS Code env. Does not print the key.
import os, json, subprocess, sys

def gemini_key():
    key = os.environ.get('GEMINI_API_KEY') or os.environ.get('GOOGLE_API_KEY') or ''
    if key.strip():
        return key.strip()
    try:
        from google.colab import userdata
        return (userdata.get('GEMINI_API_KEY') or '').strip()
    except Exception:
        return ''

key = gemini_key()
if not key:
    print(json.dumps({
        'gemini': 'skipped',
        'colab': 'Secrets → name GEMINI_API_KEY → allow notebook access',
        'vscode': 'set GEMINI_API_KEY, then run this cell or notebooks/gemini_review.py',
    }, indent=2))
else:
    os.environ['GEMINI_API_KEY'] = key
    try:
        from google import genai
    except ImportError:
        subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', '-U', 'google-genai'])
        from google import genai
    packet = {'measured': measured, 'telemetry': telemetry, 'claim': claim}
    prompt = (
        'Review this Bio 2.0 packet. Keep measured, telemetry, and author_claim separate. '
        'Do not call eta_raw 30000 an enzyme assay. Do not call the NS watch a Clay proof. '
        'German, at most 120 words, then one next notebook step inside those rules.\n'
        + json.dumps(packet)
    )
    try:
        client = genai.Client()
        interaction = client.interactions.create(model='gemini-3.8-flash', input=prompt)
        print(interaction.output_text)
    except Exception as e:
        print(json.dumps({'gemini': 'error', 'type': type(e).__name__}))
```
